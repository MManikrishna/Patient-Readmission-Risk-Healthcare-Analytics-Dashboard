```python
missing_counts = (df == "?").sum()

missing_counts[missing_counts > 0].sort_values(ascending=False)
```

approach

We will not simply delete rows containing ?.

Instead:

weight → likely exclude from our project because almost the entire column is missing.
medical_specialty → keep, convert ? → "Unknown".
payer_code → keep/transform, convert ? → "Unknown".
race → convert ? → "Unknown".
diag_1, diag_2, diag_3 → keep, handle ? as "Unknown".
Why?

If we deleted every row containing ?, we'd unnecessarily lose a large amount of useful patient data.

For example, weight has:

98,569 / 101,766 ≈ 96.9% missing.

So there's very little value in keeping it for this project.

```python
import pandas as pd

# ============================================================
# 1. Create a clean copy
# ============================================================

df_clean = df.copy()


# ============================================================
# 2. Replace '?' with 'Unknown'
# ============================================================

unknown_cols = [
    "race",
    "payer_code",
    "medical_specialty",
    "diag_1",
    "diag_2",
    "diag_3"
]

for col in unknown_cols:
    df_clean[col] = df_clean[col].replace("?", "Unknown")


# ============================================================
# 3. Standardize gender
# ============================================================

df_clean["gender"] = df_clean["gender"].replace(
    "Unknown/Invalid", "Unknown"
)


# ============================================================
# 4. Remove weight
#    ~97% values are missing
# ============================================================

df_clean = df_clean.drop(columns=["weight"])


# ============================================================
# 5. Handle glucose and A1C missing values
# ============================================================

df_clean["max_glu_serum"] = df_clean["max_glu_serum"].fillna(
    "Not Tested"
)

df_clean["A1Cresult"] = df_clean["A1Cresult"].fillna(
    "Not Tested"
)


# ============================================================
# 6. Create 30-day readmission target
# ============================================================

df_clean["readmitted_30_days"] = df_clean["readmitted"].map({
    "<30": "Yes",
    ">30": "No",
    "NO": "No"
})


# ============================================================
# 7. Create clean age group
# ============================================================

df_clean["age_group"] = (
    df_clean["age"]
    .str.replace("[", "", regex=False)
    .str.replace(")", "", regex=False)
)


# ============================================================
# 8. Create Length of Stay groups
# ============================================================

df_clean["los_group"] = pd.cut(
    df_clean["time_in_hospital"],
    bins=[0, 2, 4, 7, 14],
    labels=[
        "1-2 days",
        "3-4 days",
        "5-7 days",
        "8-14 days"
    ],
    include_lowest=True
)


# ============================================================
# 9. Create outpatient utilization group
# ============================================================

df_clean["outpatient_visit_group"] = pd.cut(
    df_clean["number_outpatient"],
    bins=[-1, 0, 2, float("inf")],
    labels=[
        "0 visits",
        "1-2 visits",
        "3+ visits"
    ]
)


# ============================================================
# 10. Create emergency utilization group
# ============================================================

df_clean["emergency_visit_group"] = pd.cut(
    df_clean["number_emergency"],
    bins=[-1, 0, 1, float("inf")],
    labels=[
        "0 visits",
        "1 visit",
        "2+ visits"
    ]
)


# ============================================================
# 11. Create inpatient utilization group
# ============================================================

df_clean["inpatient_visit_group"] = pd.cut(
    df_clean["number_inpatient"],
    bins=[-1, 0, 1, float("inf")],
    labels=[
        "0 visits",
        "1 visit",
        "2+ visits"
    ]
)


# ============================================================
# 12. Remove extremely sparse medication columns
# ============================================================

medications_to_remove = [
    "repaglinide",
    "nateglinide",
    "chlorpropamide",
    "glimepiride",
    "acetohexamide",
    "glipizide",
    "glyburide",
    "tolbutamide",
    "pioglitazone",
    "rosiglitazone",
    "acarbose",
    "miglitol",
    "troglitazone",
    "tolazamide",
    "examide",
    "citoglipton",
    "glyburide-metformin",
    "glipizide-metformin",
    "glimepiride-pioglitazone",
    "metformin-rosiglitazone",
    "metformin-pioglitazone"
]

df_clean = df_clean.drop(
    columns=medications_to_remove
)


# ============================================================
# 13. Standardize column names
# ============================================================

df_clean.columns = (
    df_clean.columns
    .str.lower()
    .str.strip()
    .str.replace("-", "_", regex=False)
    .str.replace(" ", "_", regex=False)
)


# ============================================================
# 14. Final validation
# ============================================================

print("Shape:", df_clean.shape)

print("\nDuplicate rows:", df_clean.duplicated().sum())

print("\nMissing values:")
print(df_clean.isna().sum()[df_clean.isna().sum() > 0])

print("\n30-day readmission:")
print(df_clean["readmitted_30_days"].value_counts())

print("\nAge groups:")
print(df_clean["age_group"].value_counts().sort_index())

print("\nLength of stay groups:")
print(df_clean["los_group"].value_counts().sort_index())

print("\nFirst 5 rows:")
display(df_clean.head())
```

```python
# ============================================================
# FINAL DATA VALIDATION
# ============================================================

print("1. Dataset shape:")
print(df_clean.shape)

print("\n2. Duplicate rows:")
print(df_clean.duplicated().sum())

print("\n3. Duplicate encounter IDs:")
print(df_clean["encounter_id"].duplicated().sum())

print("\n4. Missing values:")
missing = df_clean.isna().sum()
print(missing[missing > 0])

print("\n5. 30-day readmission:")
print(df_clean["readmitted_30_days"].value_counts())

print("\n6. Gender:")
print(df_clean["gender"].value_counts())

print("\n7. Race:")
print(df_clean["race"].value_counts())

print("\n8. Age groups:")
print(df_clean["age_group"].value_counts().sort_index())

print("\n9. Length of stay:")
print(df_clean["los_group"].value_counts().sort_index())

print("\n10. Data types:")
print(df_clean.dtypes)
```

cleaning is completed now 

```python
# ============================================================
# EXPORT CLEANED DATASET
# ============================================================

# Convert categorical columns to string/object
# so PostgreSQL imports them cleanly
category_cols = [
    "los_group",
    "outpatient_visit_group",
    "emergency_visit_group",
    "inpatient_visit_group"
]

for col in category_cols:
    df_clean[col] = df_clean[col].astype(str)

# Save cleaned dataset
output_path = "healthcare_readmission_cleaned.csv"

df_clean.to_csv(
    output_path,
    index=False
)

print("Cleaned dataset exported successfully.")
print("File:", output_path)
print("Shape:", df_clean.shape)
```
